```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import h5py
import warnings
warnings.filterwarnings('ignore')
```

```python
datas=pd.read_csv('output.csv')
```

6 - creating the object for CountVectorizer

```python
cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
# Replace NaN values with "unknown"
datas['genres'].fillna("", inplace=True)
```

8,9 --9th screenshot

```python
feature_vectors_genres= cv.fit_transform(datas['genres']).toarray()
```

```python
feature_vectors_genres[0]
```

10,11--11th screenshot

```python
feature_vectors_overview= cv.fit_transform(datas['overview']+datas['keywords']).toarray()
```

```python
feature_vectors_overview[0]
```

```python
feature_vectors_language= cv.fit_transform(datas['spoken_languages']).toarray()
```

```python
feature_vectors_language[0]
```

```python
# Replace NaN values with "unknown"
datas['cast'].fillna("", inplace=True)
datas['crew'].fillna("", inplace=True)
```

15,16--16th screenshot

```python
feature_vectors_credits= cv.fit_transform(datas['cast']+datas['crew']).toarray()
```

```python
feature_vectors_credits[0]
```

17,19,20,21--18th screenshot

```python
similarity_genres = cosine_similarity(feature_vectors_genres)
```
